```python
from transformers import AutoModelForCausalLM, AutoTokenizer, pipeline
from langchain_community.llms import HuggingFacePipeline
from langchain.chains import LLMChain
from langchain.prompts import PromptTemplate
```

```python
template = "Translate the following text to Spanish: {text}"
prompt = PromptTemplate.from_template(template)
```

```python
model_id = "gpt2"  # pick a small model; swap for any chat/instruct model you like
tokenizer = AutoTokenizer.from_pretrained(model_id)
model = AutoModelForCausalLM.from_pretrained(model_id)
gen_pipe = pipeline("text-generation", model=model, tokenizer=tokenizer, max_new_tokens=64)
```

```python
llm = HuggingFacePipeline(pipeline=gen_pipe)
chain = LLMChain(llm=llm, prompt=prompt)
print(chain.run("Hello world"))
```
